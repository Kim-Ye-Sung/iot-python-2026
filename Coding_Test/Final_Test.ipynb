# 주제 : 음식감지 AI

- 음식 사진을 넣는다.
- YOLO 모델로 이미지를 분석한다.
- 감지된 음식 이름을 출력한다.
- 결과 이미지를 저장한다.

## 1. 라이브러리 불러오기

```python
from ultralytics import YOLO
```

## 2. YOLO 모델 불러오기

```python
model = YOLO('yolo11n.pt')
```

## 3. 감지할 음식 목록 만들기

YOLO 기본 모델은 모든 음식을 인식하지는 못한다. 따라서 기본 모델이 인식 가능한 간단한 음식만 감지한다.

```python
food_list = [
    'banana',
    'apple',
    'sandwich',
    'orange',
    'broccoli',
    'carrot',
    'hot dog',
    'pizza',
    'donut',
    'cake'
]
```

## 4. 감지된 음식 출력하기

```python
food_image = './pizza.jpg'  # 감지할 음식 사진 지정. 여기에 다른 음식 사진 넣으면 됩니다.

results = model(food_image)  

detected_foods = []

for box in results[0].boxes:
    cls_id = int(box.cls[0])
    label = model.names[cls_id]

    if label in food_list:
        detected_foods.append(label)

if len(detected_foods) > 0:
    print('감지된 음식')

    for food in set(detected_foods):
        print('-', food)
else:
    print('감지된 음식이 없습니다.')
```

## 5.결과 이미지 확인하기

```python
results[0].show()
```

## 6. 결과 이미지 저장하기 ("food_image 변수 + _result".jpg로 저장)

```python
file_name = food_image.split('/')[-1]
file_name = file_name.split('.')[0]

save_name = file_name + '_result.jpg'

results[0].save(filename=save_name)
```
